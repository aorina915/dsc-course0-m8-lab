```python
injury_cols = ['Total.Fatal.Injuries', 'Total.Serious.Injuries', 'Total.Minor.Injuries', 'Total.Uninjured']

for col in injury_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0).astype(int)


df["Total.occupants"] = df[injury_cols].sum(axis=1)


#Fatal injuries

df['Fatal.Serious.Count'] = (
    df['Total.Serious.Injuries'].fillna(0) + 
    df['Total.Fatal.Injuries'].fillna(0)
)


df['Fatal.Serious.Rate'] = (
    df['Fatal.Serious.Count'] / df['Total.occupants']
).where(df['Total.occupants'] > 0)



df['Fatal.Serious.Flag'] = (
    df['Fatal.Serious.Count'] > 0
).astype('boolean')



print(df[
    ['Total.occupants', 'Fatal.Serious.Count', 'Fatal.Serious.Rate', 'Fatal.Serious.Flag']
].head())
```

### Cleaning other columns
- there are other columns containing data that might be related to the outcome of an accident. We list a few here:
- Engine.Type
- Weather.Condition
- Number.of.Engines
- Purpose.of.flight
- Broad.phase.of.flight

Inspect and identify potential cleaning tasks in each of the above columns. Execute those cleaning tasks. 

**Note**: You do not necessarily need to impute or drop NaNs here.

```python
# Engine type
df["Engine.Type"] = df["Engine.Type"].astype(str).str.strip()
df["Engine.Type"] = df["Engine.Type"].replace({"nan": "Unknown", "": "Unknown"})

# Weather condition
df["Weather.Condition"] = df["Weather.Condition"].astype(str).str.strip()
df["Weather.Condition"] = df["Weather.Condition"].replace({"nan": "Unknown", "": "Unknown"})

# Engine count
df["Number.of.Engines"] = pd.to_numeric(df["Number.of.Engines"], errors="coerce")
df["Number.of.Engines"] = df["Number.of.Engines"].fillna(df["Number.of.Engines"].median())

# Purpose of flight
df["Purpose.of.flight"] = df["Purpose.of.flight"].astype(str).str.strip()
df["Purpose.of.flight"] = df["Purpose.of.flight"].replace({"nan": "Unknown", "": "Unknown"})

# Flight phase
df["Broad.phase.of.flight"] = df["Broad.phase.of.flight"].astype(str).str.strip()
df["Broad.phase.of.flight"] = df["Broad.phase.of.flight"].replace({"nan": "Unknown", "": "Unknown"})
```
